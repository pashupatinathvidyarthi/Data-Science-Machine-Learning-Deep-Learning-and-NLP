## Numpy 

Numpy is a fundamental library for scientific computing in Python. It provides support for arrays and matrices, along with a collection of mathematical functions to operate on these data sstrructures. In this lesson, we will cover the basics of Numpy, focusing on arrays and vectorized operations.

```python
import numpy as np

##create an array using numpy
##create 1D array
arr1=np.array([1,2,3,4,5,6])
print(arr1)
print(type(arr1))
print(arr1.shape)
```

```python
## Reshaping to 2D
arr2=np.array([1,2,3,4,5])
arr2.reshape(1,5)
```

```python
arr2=np.array([[1,2,3,4,5]])
arr2.shape
```

```python
##2D array
arr2=np.array([[1,2,3,4,5],[2,3,4,5,6]])
print(arr2)
print(arr2.shape)
```

```python
np.arange(0,10,2)
```

```python
##Convert to 2D array
np.arange(0,10,2).reshape(5,1)
```
