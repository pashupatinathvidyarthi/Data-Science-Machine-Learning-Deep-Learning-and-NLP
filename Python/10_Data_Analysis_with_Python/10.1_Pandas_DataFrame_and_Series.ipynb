## Pandas-DataFrame and Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures. Series and DataFrame. A Series is a one-dimensional array-like object, while DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
## Series
## A Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table.

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
```

```python
## Create a series from dictionary 
data={'a':1,'b':2,'c':3,'d':4}
series=pd.Series(data)
print(series)
```

```python
## To give specific index
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
##DataFrame
## create a dataframe from a dictionary of lisst

data={
    "Name":['Pashupati','Vantika','Abhishek'],
    "Age":[23,5,25],
    "City":["Bikramganj","Arrah","Bikramganj"]
}

df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Creating New DataFrame

import pandas as pd

data={
    "Name" : ["Ram", "Shyam", "Mohan"],
    "Age" : [20, 21, 22],
    "City" : ["Delhi", "Patna", "Bhopal"]
}

df1=pd.DataFrame(data)
```

```python
## Data Manipulation with DataFrame

df1
```

```python
## Adding a new column
df1["New City"]=["Kolkata", "Bangalor", "Gujarat"]
df1
```

```python
## Remove a column
## drop is temporary operation
##For permanently remove we have to use inplace =True
df1.drop("New City", axis=1)   ##when axis=0 it check row and when axis=1 it check column
df1
```

```python
## drop is temporary operation
##For permanently remove we have to use inplace =True
df1.drop("New City",axis=1,inplace=True)
df1
```

```python
## Add age to the columns
df1['Age']=df1['Age']+1
df1
```

```python
df1.drop(0) ##Drop based on row
##This again not a permanent operation
```

```python
df1
```

```python
## Now this permanent Operation
df1.drop(0, inplace=True)
```
