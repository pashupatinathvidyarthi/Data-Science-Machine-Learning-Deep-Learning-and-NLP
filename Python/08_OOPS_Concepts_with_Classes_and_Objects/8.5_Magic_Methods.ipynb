## Magic Methods

Magic methods in Python, also known as dunder methods (double underscore methods), are special methods that starts and end with double underscores. These methods enable you to define the behaviour of objects for built-in operations, such as arithmetic operations, comparisons,a nd more.

## Magic Methods 

Magic methods are predefined methods in Python that you can override to change the behaviour of your objects. Some common magic methods include:



```python
"""
__init__ initializes a new instance of a class.
__str__ Return a string representation of an object.
__repr__ Returns an official string representation of an object.
__len__ Returns the length of an object.
__getitem__ Gets an item from a container
__setitem__ Sets an item in a container
"""
```

```python
## Basci Magic Methods
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
person=Person("Pashupati",23)
print(person)
```

```python
## Now we are going to over-write the output of above code

## Basci Magic Methods
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self): ##__str__ here to use to over write the message
        return f"{self.name},{self.age} years old"
    
person=Person("Pashupati",23)
print(person)
```
