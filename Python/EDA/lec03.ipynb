```python
df_copy['Size']=df_copy['Size'].str.replace('M','000')
df_copy['Size']=df_copy['Size'].str.replace('k','')
df_copy['Size']=df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size']=df_copy['Size'].astype(float)
```

```python
chars_to_remove=['+',',','$']
cols_to_clean=['Installs','Price']
for item in chars_to_remove:
    for cols in cols_to_clean:
        df_copy[cols]=df_copy[cols].str.replace(item,'')
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Installs']=df_copy['Installs'].astype(int)
df_copy['Price']=df_copy['Price'].astype(float)
```

```python
df_copy.info()
```

```python
## Handling Last Updated Column
df_copy['Last Updated'].unique()
```

```python
pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])
df_copy['Day']=df_copy['Last Updated'].dt.day
df_copy['Month']=df_copy['Last Updated'].dt.month
df_copy['Year']=df_copy['Last Updated'].dt.year
```

## Explore Data

```python
numeric_features=[feature for feature in df_copy.columns if df_copy[feature].dtype!='O']
categorical_features=[feature for feature in df_copy.columns if df_copy[feature].dtype=='O']

# print columns
print('We have {} numerical features : {}'.format(len(numeric_features),numeric_features))
print('We have {} categorical features : {}'.format(len(categorical_features),categorical_features))
```

## 3.2 Feature Information
1. App :- Name of the App
2. Category :- Category to which the App falls.
3. Rating : Applications's rating on playstore
4. Reviews : Number of reviews of the app.
5. Size : Size of the app.
6. Installs : Number of installs of the app
7. Type : If the app is free/paid
8. Price : Price of the app (o if it is free)
9. Content Rating : Appropriate Target Auidence of the app
10. Genres : Genre under which the app falls.
11. Last Updated : Date on which the app was last updated.
12. Current Ver : Current version of the app
13. Android Ver : Minimum required Android version to run the app

```python
## Proportion of count data on categorical features
for col in categorical_features:
    print(df[col].value_counts(normalize=True)*100)
    print('-----------------------------')
```

```python
## Proportion of count data on numerical columns
plt.figure(figsize=(15,15))
plt.suptitle('Univariate Analysis of Numerical Features',fontsize=20,fontweight='bold',alpha=0.8,y=1)

for i in range(0,len(numeric_features)):
    plt.subplot(5,3,i+1)
    sns.kdeplot(x=df_copy[numeric_features[i]],shade=True,color='r')
    ## kdeplot kernal density estimation plot
    plt.xlabel(numeric_features[i])
    plt.tight_layout()
```

## Obseervations
- Rating and Year is left skewed while Reviews,Size and Installs are right skewed

```python
# Categorical columns
plt.figure(figsize=(12, 8))
plt.suptitle('Univariate Analysis of Categorical Features',fontsize=20, fontweight='bold', alpha=0.8,y=1)
category=['Type','Content Rating']

for i in range(0,len(category)):
    plt.subplot(2, 2,i + 1)
    sns.countplot(x=df[category[i]], palette='Set2')
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
