### EDA and Feature Engineering - Flight Price Prediction

Check the dataset here:  
https://www.kaggle.com/datasets/shubhambathwal/flight-price-prediction

---

#### Features

The various features of the cleaned dataset are explained below:

1. **Airline**  
   The name of the airline company.  
   - Type: Categorical  
   - Unique values: 6 airlines  

2. **Flight**  
   Stores the flight code of the plane.  
   - Type: Categorical  

3. **Source City**  
   The city from which the flight takes off.  
   - Type: Categorical  
   - Unique values: 6 cities  

4. **Departure Time**  
   A derived categorical feature created by grouping time periods into bins.  
   - Type: Categorical  
   - Unique values: 6 time labels  

5. **Stops**  
   Number of stops between source and destination.  
   - Type: Categorical  
   - Unique values: 3 (e.g., non-stop, one-stop, etc.)  

6. **Arrival Time**  
   A derived categorical feature created by grouping time intervals into bins.  
   - Type: Categorical  
   - Unique values: 6 time labels  

7. **Destination City**  
   The city where the flight will land.  
   - Type: Categorical  
   - Unique values: 6 cities  

8. **Class**  
   Information about seat class.  
   - Type: Categorical  
   - Values: Business, Economy  

9. **Duration**  
   Total travel time between cities.  
   - Type: Continuous  
   - Unit: Hours  

10. **Days Left**  
    Number of days left before departure.  
    - Type: Numerical  

---

## Target Variable

- **Price**  
  The cost of the flight ticket.  
  - Type: Continuous (Regression Problem)
